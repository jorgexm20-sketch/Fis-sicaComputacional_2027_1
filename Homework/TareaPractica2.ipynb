# Constante  de Madelung

Primero construimos nuestros arreglos array usando un valor de L, cada una tendrá 2*L + 1 elementos.

Después vamos a crear las combinaciones para i,j,k

```python
L = 100 # Valor de nuestra L 

lista_i = np.arange(-L, L+1) # Creamos las listas
lista_j = np.arange(-L, L+1)
lista_k = np.arange(-L, L+1)
```

```python
M = 0  # Creamos M  = 0 para que se vaya acumulando los valores de 1/ Norma del vector
M_teorica = 1.74756
# Tenemos que crear For anidados, esto nos va a costar bastante en poder computacional 
for i in lista_i:
    for j in lista_j:
        for k in lista_k:
            if i == 0 and j == 0 and k == 0: 
                continue # El vector no puede ser el origen porque la norma es cero y se divide entre cero
            vector = np.array([i,j,k]) # Creamos cada vector con entradas i,j,k de nuestros arreglos 
            norma = np.linalg.norm(vector) #Sacamos la norma del vector
            P = i + j + k # Necesitamos condiciones
            if P % 2 == 0:
                M += 1/norma
            else:
                M -= 1/norma

print(f"Para L = {L} tenemos que M = {M} y M_teórica = {M_teorica}")           
            

```

# 3. Coeficientes binomiales

## A) función binomial(n,k)

```python
def factorial_recursivo(n):
    n = int(n)
    if n == 0:
        return 1
    elif n>0:
        return n * factorial_recursivo(n-1)
```

```python
def binomial(n,k):
    #if k == 0:
    #if k>= 1:
    coeficiente_binomial = (factorial_recursivo(n))/(factorial_recursivo(k)*factorial_recursivo(n-k))
    return int(coeficiente_binomial)
```

## Cálculo de probabilidad 

### 1- La probabilidad de que para una moneda no sesgada, lanzada n= 100 veces, salga águila k = 60 veces 

```python
# Definimos las variables
N = 100
K = 60
probabilidad_N_K = binomial(N,K)/(2**N)
print(f"la probabilidad de lanzar {N} veces la moneda y que caiga {K} águila es {probabilidad_N_K}")
```

### La probabilidad de que salga aguila 60 veces o más.

```python
N = 100
K = 60
probabilidad_N_K_60 = 0 
for i in range(K, N+1):
    probabilidad_N_K_60 += binomial(N,i)/(2**N)

print(f"la probabilidad de lanzar {N} veces la moneda y que caiga {K} veces o más águila es {probabilidad_N_K_60}")
```

# 4. Números primos

## números primos hasta 10 000 (diez mil).

```python
# Creamos nuestras variables
primos = [2] 
N = 10000
for n in range(2,N+1): # nuestro rango es 1000
    es_primo = True # Cremaos una variable suponiendo que n es primo 

    for i in primos: # Veamos si es divisible entre otro número primo 
        if n % i == 0: 
            es_primo = False # si es divisible entonces no es primo 
    # Lo agregamos a la lista de primos 
    if es_primo:
        primos.append(n)
```

# 5. Recursión

## a) usando recursividad, que calcule Cn en la serie de Catalan

```python
def Catalan(n):
    n = int(n)
    if n< 0:
        return print(f"Tu numero n debe ser positivo {n}")
    if n == 0:
        return 1
    if n>0:
        cn = ((4*n-2)/(n+1))*Catalan(n-1)
        return int(cn)
```

## Emplee recursividad para calcular el maximo común divisor de m y n.

```python
def g(m,n):
    m = int(m)
    n = int(n)
    if m<0 or n<0:
        return print(f"Tu número m = {m} o n = {n} deben ser positivos")
    if n == 0:
        return m 
    if n>0:
        return g(n, m%n)
```
