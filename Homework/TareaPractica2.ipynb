## 3- Coeficientes binomiales 


```python
import numpy as np
import math
```

```python
def binomial(n,k):
    
    if k == 0:
        return 1 
    if k>= 1 and n>k:
        coeficiente_binomial = (math.factorial(n))/(math.factorial(k)*math.factorial(n-k))
        return int(coeficiente_binomial)
```

```python
binomial(15,6)
```

```python
for i in range(1,50):
    for j in range(1,15):
        if i>j:

            print(f"el binomial calculado de {i} y {j} es {binomial(i,j)}")
```
