```python
"""
El usuario tiene que poner dos vectores
para cada entrada del vector:
    se multiplica entrada por entrada y se guarda en una suma
"""
lista_palabras = ["primera" , "segunda", "tercera"]
vector_1 = []
vector_2 = []
for i in lista_palabras:
    vector_1.append(float(input(f"La {i} entrada de tu primer vector: ")))

for i in lista_palabras:
    vector_2.append(float(input(f"La {i} entrada de tu segundo vector: ")))
print("Tu vectores son: ")
print(vector_1)
print(vector_2)

producto_escalar = 0
for i in range(len(vector_1)):
    producto_escalar += vector_1[i] * vector_2[i]
print(f"Tu producto escalar es: {producto_escalar}")
```

```python
"""
El usuario tiene que poner dos vectores de n dimensiones
lo guardamos en una lista
para cada entrada del vector:
    se multiplica entrada por entrada y se guarda en una suma
"""
n = int(input("¿De qué dimensión quieres que sea el vector?: "))
a_1 = random.sample(range(2*n), n)
b_1 = random.sample(range(2*n), n)
print("Tus vectores son: ")
print(a_1)
print(b_1)

producto_escalar = 0
for i in range(n):
    producto_escalar += a_1[i] * b_1[i]
print(f"Tu producto escalar es: {producto_escalar}")
```
