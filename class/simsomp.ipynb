```python
def IntegraSimsom(f, N, b,a):
    h = (b-a)/N
    suma_par = 0 
    suma_impar = 0 
    for i in range(1,int((N/2))):
        suma_par += f(a+(2*i*h))
    for i in range(1,int((N/2))+1):
        suma_impar += f(a+(2*i-1)*h)
    return round((h/3)* ((f(a)+f(b))  + (4 * suma_impar) + (2 * suma_par)), 10)

```

```python
IntegraSimsom(cuad, 1000,10,0)
```
